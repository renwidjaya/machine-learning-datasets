```
# ============================================================
# P3 - MARKETING CAMPAIGN CUSTOMER SEGMENTATION
# Nama  : Rendi Widjaya
# NIM   : 56226110048
# Algoritma: K-Means Clustering
# ============================================================


# ============================================================
# IMPORT LIBRARY
# ============================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

pd.set_option('display.max_columns', None)

sns.set_theme(style='whitegrid')


# ============================================================
# 1. UPLOAD DAN MEMBACA DATASET
# ============================================================
from google.colab import files

print("Silakan upload file P3_marketing_campaign.csv")

uploaded = files.upload()

# Membaca file yang di-upload
file_name = next(iter(uploaded))

# Dataset menggunakan pemisah tab (TSV)
df = pd.read_csv(
    file_name,
    sep='\t'
)

print("\n===== P3 SCREENSHOT 01 - INFORMASI DATASET =====")

print("Nama file:", file_name)
print("Ukuran dataset:", df.shape)
print("Jumlah baris:", df.shape[0])
print("Jumlah kolom:", df.shape[1])

print("\nLima data pertama (kolom terpilih):")

kolom_tampil = [
    'ID',
    'Year_Birth',
    'Education',
    'Marital_Status',
    'Income',
    'Kidhome',
    'Teenhome',
    'Recency',
    'MntWines',
    'MntMeatProducts',
    'Response'
]

display(df[kolom_tampil].head())

print("\nInformasi tipe data:")
df.info()


# ============================================================
# 2. PRE-PROCESSING DATA
# ============================================================
# 2A. PEMERIKSAAN MISSING VALUE
# ------------------------------------------------------------
print("\n===== P3 SCREENSHOT 02 - MISSING VALUE =====")

print(df.isnull().sum())

print("\nTotal missing value:")
print(df.isnull().sum().sum())


# ------------------------------------------------------------
# 2B. PEMERIKSAAN DATA DUPLIKAT
# ------------------------------------------------------------
print("\n===== P3 SCREENSHOT 03 - DATA DUPLIKAT =====")

print("Jumlah data duplikat:", df.duplicated().sum())


# ------------------------------------------------------------
# 2C. PEMERIKSAAN DATA KATEGORIKAL
# ------------------------------------------------------------
print("\n===== P3 SCREENSHOT 04 - DISTRIBUSI KATEGORIKAL =====")

categorical_columns_initial = df.select_dtypes(
    include=['object']
).columns.tolist()

for col in categorical_columns_initial:
    print("\nDistribusi", col)
    print(df[col].value_counts(dropna=False))


# ------------------------------------------------------------
# 2D. MENGHAPUS IDENTITAS DAN VARIABEL KONSTANTA
# ------------------------------------------------------------
# ID hanya merupakan identitas pelanggan.
# Z_CostContact dan Z_Revenue merupakan variabel konstan
# sehingga tidak memberikan informasi pembeda untuk clustering.

df = df.drop(
    columns=['ID', 'Z_CostContact', 'Z_Revenue']
)

print("\nKolom setelah penghapusan:")
print(df.columns.tolist())


# ============================================================
# 3. ANALISIS DESKRIPTIF DATASET
# ============================================================
print("\n===== P3 SCREENSHOT 05 - STATISTIK DESKRIPTIF =====")

display(df.describe().T)


# ------------------------------------------------------------
# 3A. DISTRIBUSI PENDIDIKAN
# ------------------------------------------------------------
print("\nDistribusi Education:")
print(df['Education'].value_counts())


# ------------------------------------------------------------
# 3B. DISTRIBUSI STATUS PERNIKAHAN
# ------------------------------------------------------------
print("\nDistribusi Marital Status:")
print(df['Marital_Status'].value_counts())


# ------------------------------------------------------------
# 3C. VISUALISASI PENDAPATAN
# ------------------------------------------------------------
print("\n===== P3 SCREENSHOT 06 - DISTRIBUSI PENDAPATAN =====")

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='Income',
    kde=True
)

plt.title('Distribusi Pendapatan Pelanggan')
plt.xlabel('Income')
plt.ylabel('Jumlah Pelanggan')
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 3D. VISUALISASI PENGELUARAN
# ------------------------------------------------------------
spending_columns = [
    'MntWines',
    'MntFruits',
    'MntMeatProducts',
    'MntFishProducts',
    'MntSweetProducts',
    'MntGoldProds'
]

print("\n===== P3 SCREENSHOT 07 - DISTRIBUSI PENGELUARAN =====")

spending_summary = df[spending_columns].sum().sort_values(
    ascending=False
)

display(spending_summary.to_frame(name='Total Spending'))

plt.figure(figsize=(10, 6))

sns.barplot(
    x=spending_summary.values,
    y=spending_summary.index
)

plt.title('Total Pengeluaran Berdasarkan Kategori Produk')
plt.xlabel('Total Pengeluaran')
plt.ylabel('Kategori Produk')
plt.tight_layout()
plt.show()


# ============================================================
# 4. FEATURE ENGINEERING
# ============================================================
# Membuat fitur total pengeluaran pelanggan
df['Total_Spending'] = df[spending_columns].sum(axis=1)

# Membuat fitur total pembelian berdasarkan kanal
purchase_columns = [
    'NumWebPurchases',
    'NumCatalogPurchases',
    'NumStorePurchases'
]

df['Total_Purchases'] = df[purchase_columns].sum(axis=1)

# Membuat fitur total anak dalam rumah tangga
df['Total_Children'] = df['Kidhome'] + df['Teenhome']

print("\n===== P3 SCREENSHOT 08 - FEATURE ENGINEERING =====")

display(
    df[
        [
            'Total_Spending',
            'Total_Purchases',
            'Total_Children'
        ]
    ].head()
)


# ============================================================
# 5. MENENTUKAN FITUR CLUSTERING
# ============================================================
# Variabel numerik untuk clustering
numeric_features = [
    'Year_Birth',
    'Income',
    'Kidhome',
    'Teenhome',
    'Total_Children',
    'Recency',
    'MntWines',
    'MntFruits',
    'MntMeatProducts',
    'MntFishProducts',
    'MntSweetProducts',
    'MntGoldProds',
    'Total_Spending',
    'NumDealsPurchases',
    'NumWebPurchases',
    'NumCatalogPurchases',
    'NumStorePurchases',
    'Total_Purchases',
    'NumWebVisitsMonth'
]

# Variabel kategorikal untuk clustering
categorical_features = [
    'Education',
    'Marital_Status'
]

# Menggabungkan fitur yang digunakan
selected_features = (
    numeric_features + categorical_features
)

X = df[selected_features].copy()

print("\n===== P3 SCREENSHOT 09 - FITUR CLUSTERING =====")

print("Jumlah fitur numerik:", len(numeric_features))
print("Jumlah fitur kategorikal:", len(categorical_features))
print("Jumlah seluruh fitur:", len(selected_features))

display(X.head())


# ============================================================
# 6. PREPROCESSING DAN STANDARDISASI
# ============================================================
# Preprocessing numerik:
# - Missing value diisi menggunakan median
# - Data distandardisasi menggunakan StandardScaler

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])


# Preprocessing kategorikal:
# - Missing value diisi menggunakan modus
# - Data diubah menggunakan OneHotEncoder
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])


# Menggabungkan preprocessing numerik dan kategorikal
preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])


# Melakukan transformasi data
X_processed = preprocessor.fit_transform(X)

# Mengubah sparse matrix menjadi dense matrix jika diperlukan
if hasattr(X_processed, 'toarray'):
    X_processed = X_processed.toarray()


print("\n===== P3 SCREENSHOT 10 - HASIL STANDARDISASI =====")

print("Ukuran data sebelum preprocessing:", X.shape)
print("Ukuran data setelah preprocessing:", X_processed.shape)

print("\nContoh data hasil preprocessing:")
display(pd.DataFrame(X_processed).head())


# ============================================================
# 7. MENENTUKAN JUMLAH CLUSTER
# ============================================================
# Menggunakan Elbow Method dan Silhouette Score
inertia_values = []
silhouette_values = []

k_values = range(2, 9)

for k in k_values:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    cluster_labels = kmeans.fit_predict(X_processed)

    inertia_values.append(kmeans.inertia_)

    silhouette_values.append(
        silhouette_score(
            X_processed,
            cluster_labels
        )
    )


# ------------------------------------------------------------
# 7A. ELBOW METHOD
# ------------------------------------------------------------
print("\n===== P3 SCREENSHOT 11 - ELBOW METHOD =====")

plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    inertia_values,
    marker='o'
)

plt.title('Elbow Method - Penentuan Jumlah Cluster')
plt.xlabel('Jumlah Cluster (K)')
plt.ylabel('Inertia')
plt.xticks(list(k_values))
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 7B. SILHOUETTE SCORE
# ------------------------------------------------------------
print("\n===== P3 SCREENSHOT 12 - SILHOUETTE SCORE =====")

silhouette_results = pd.DataFrame({
    'Jumlah Cluster': list(k_values),
    'Silhouette Score': silhouette_values
})

display(silhouette_results)

plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    silhouette_values,
    marker='o'
)

plt.title('Silhouette Score untuk Setiap Jumlah Cluster')
plt.xlabel('Jumlah Cluster (K)')
plt.ylabel('Silhouette Score')
plt.xticks(list(k_values))
plt.tight_layout()
plt.show()


# ============================================================
# 8. MEMBANGUN MODEL K-MEANS
# ============================================================
# Jumlah cluster awal menggunakan K=4.
# Nilai ini merupakan kandidat awal dan perlu ditinjau
# kembali berdasarkan Elbow Method dan Silhouette Score.

optimal_k = 4

kmeans_final = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans_final.fit_predict(X_processed)

# Menambahkan hasil cluster ke dataset
df['Cluster'] = cluster_labels


print("\n===== P3 SCREENSHOT 13 - HASIL CLUSTERING =====")

print("Jumlah cluster:", optimal_k)

print("\nJumlah pelanggan pada setiap cluster:")
print(df['Cluster'].value_counts().sort_index())


# ============================================================
# 9. VISUALISASI CLUSTER
# ============================================================
# Menggunakan PCA untuk mengubah data menjadi dua dimensi
# agar hasil clustering dapat divisualisasikan.

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_processed)

df['PCA1'] = X_pca[:, 0]
df['PCA2'] = X_pca[:, 1]


print("\n===== P3 SCREENSHOT 14 - VISUALISASI CLUSTER =====")

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=df,
    x='PCA1',
    y='PCA2',
    hue='Cluster',
    palette='Set2',
    s=50
)

plt.title('Visualisasi Customer Segmentation Menggunakan K-Means')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(title='Cluster')
plt.tight_layout()
plt.show()


# ============================================================
# 10. PROFILING CLUSTER
# ============================================================
# Menghitung jumlah pelanggan pada setiap cluster
print("\n===== P3 SCREENSHOT 15 - JUMLAH PELANGGAN PER CLUSTER =====")

cluster_size = df['Cluster'].value_counts().sort_index()

display(
    cluster_size.to_frame(name='Jumlah Pelanggan')
)


# ------------------------------------------------------------
# 10A. RATA-RATA KARAKTERISTIK SETIAP CLUSTER
# ------------------------------------------------------------
profiling_columns = [
    'Year_Birth',
    'Income',
    'Total_Children',
    'Recency',
    'Total_Spending',
    'Total_Purchases',
    'NumDealsPurchases',
    'NumWebVisitsMonth',
    'NumWebPurchases',
    'NumCatalogPurchases',
    'NumStorePurchases'
]

print("\n===== P3 SCREENSHOT 16 - PROFILING CLUSTER =====")

cluster_profile = df.groupby('Cluster')[
    profiling_columns
].mean().round(2)

display(cluster_profile)


# ------------------------------------------------------------
# 10B. DISTRIBUSI KATEGORI SETIAP CLUSTER
# ------------------------------------------------------------
print("\n===== P3 SCREENSHOT 17 - KARAKTERISTIK KATEGORIKAL =====")

for col in categorical_features:

    print("\nDistribusi", col, "berdasarkan Cluster:")

    display(
        pd.crosstab(
            df['Cluster'],
            df[col],
            normalize='index'
        ).round(3) * 100
    )


# ============================================================
# 11. VISUALISASI PROFILING CLUSTER
# ============================================================
print("\n===== P3 SCREENSHOT 18 - PERBANDINGAN PENGELUARAN CLUSTER =====")

cluster_spending = df.groupby('Cluster')[
    spending_columns
].mean()

display(cluster_spending.round(2))

cluster_spending.plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title('Rata-rata Pengeluaran Produk Berdasarkan Cluster')
plt.xlabel('Cluster')
plt.ylabel('Rata-rata Pengeluaran')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ============================================================
# 12. EVALUASI HASIL CLUSTERING
# ============================================================
final_silhouette = silhouette_score(
    X_processed,
    cluster_labels
)

print("\n===== P3 SCREENSHOT 19 - EVALUASI CLUSTERING =====")

print("Jumlah Cluster:", optimal_k)
print("Silhouette Score:", round(final_silhouette, 4))

print("\nProporsi pelanggan setiap cluster:")

cluster_percentage = (
    df['Cluster'].value_counts(normalize=True)
    .sort_index() * 100
)

display(
    cluster_percentage.to_frame(
        name='Persentase Pelanggan'
    ).round(2)
)


# ============================================================
# SELESAI - P3 MARKETING CAMPAIGN CLUSTERING
# ============================================================
```
