```
# ============================================================
# P2 - BANK CUSTOMER CHURN PREDICTION
# Nama  : Rendi Widjaya
# NIM   : 56226110048
# ============================================================

# ============================================================
# IMPORT LIBRARY
# ============================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

# ============================================================
# 1. MEMBACA DATASET
# ============================================================
df = pd.read_csv('/content/P2_BankChurners.csv')

print("\n===== P2 SCREENSHOT 01 - INFORMASI DATASET =====")
print("Ukuran dataset:", df.shape)
display(df.head())


# ============================================================
# 2. PRE-PROCESSING DATA
# ============================================================
# Menghapus kolom kosong dan kolom identitas nasabah
df = df.drop(columns=['Unnamed: 21', 'CLIENTNUM'])

# ------------------------------------------------------------
# 2A. PEMERIKSAAN MISSING VALUE
# ------------------------------------------------------------
print("\n===== P2 SCREENSHOT 02 - MISSING VALUE =====")
print(df.isnull().sum())

# ------------------------------------------------------------
# 2B. PEMERIKSAAN DATA DUPLIKAT
# ------------------------------------------------------------
print("\n===== P2 SCREENSHOT 03 - DATA DUPLIKAT =====")
print("Jumlah data duplikat:", df.duplicated().sum())

# ------------------------------------------------------------
# 2C. IDENTIFIKASI VARIABEL TARGET
# ------------------------------------------------------------
print("\n===== P2 SCREENSHOT 04 - DISTRIBUSI TARGET =====")

print("Jumlah setiap kelas:")
print(df['Attrition_Flag'].value_counts())

print("\nPersentase setiap kelas:")
print(
    (df['Attrition_Flag'].value_counts(normalize=True) * 100).round(2)
)

# Memisahkan variabel independen dan dependen
X = df.drop(columns=['Attrition_Flag'])

# Mengubah target menjadi numerik:
# 0 = Existing Customer
# 1 = Attrited Customer

y = (df['Attrition_Flag'] == 'Attrited Customer').astype(int)

# Mengidentifikasi kolom kategorikal dan numerik
categorical_columns = X.select_dtypes(
    include=['object']
).columns.tolist()

numeric_columns = X.select_dtypes(
    exclude=['object']
).columns.tolist()

print("\nKolom kategorikal:")
print(categorical_columns)

print("\nKolom numerik:")
print(numeric_columns)


# ============================================================
# 3. ANALISIS DESKRIPTIF DATASET
# ============================================================
# 3A. STATISTIK DESKRIPTIF NUMERIK
# ------------------------------------------------------------
print("\n===== P2 SCREENSHOT 05 - STATISTIK NUMERIK =====")
display(df[numeric_columns].describe().T)

# ------------------------------------------------------------
# 3B. STATISTIK DESKRIPTIF KATEGORIKAL
# ------------------------------------------------------------
print("\n===== P2 SCREENSHOT 06 - STATISTIK KATEGORIKAL =====")
display(df.describe(include='object').T)

# ------------------------------------------------------------
# 3C. VISUALISASI DISTRIBUSI TARGET
# ------------------------------------------------------------
print("\n===== P2 SCREENSHOT 07 - GRAFIK DISTRIBUSI TARGET =====")

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x='Attrition_Flag'
)

plt.title('Distribusi Status Nasabah')
plt.xlabel('Status Nasabah')
plt.ylabel('Jumlah')
plt.tight_layout()
plt.show()


# ============================================================
# 4. ANALISIS HUBUNGAN VARIABEL X DAN Y
# ============================================================
# Membuat salinan variabel numerik
df_corr = df[numeric_columns].copy()

# Menambahkan target numerik
df_corr['Attrition_Target'] = y

# Menghitung korelasi setiap variabel terhadap target
correlation = (
    df_corr.corr()['Attrition_Target']
    .drop('Attrition_Target')
    .sort_values(key=abs, ascending=False)
)

# ------------------------------------------------------------
# 4A. TABEL KORELASI
# ------------------------------------------------------------
print("\n===== P2 SCREENSHOT 08 - KORELASI TARGET =====")
display(correlation.to_frame(name='Correlation'))

# ------------------------------------------------------------
# 4B. VISUALISASI KORELASI
# ------------------------------------------------------------
print("\n===== P2 SCREENSHOT 09 - GRAFIK KORELASI =====")

plt.figure(figsize=(10, 7))

sns.barplot(
    x=correlation.values,
    y=correlation.index
)

plt.title('Korelasi Variabel Numerik terhadap Customer Churn')
plt.xlabel('Nilai Korelasi')
plt.ylabel('Variabel')
plt.tight_layout()
plt.show()


# ============================================================
# 5. PEMBAGIAN DATA TRAINING DAN TESTING
# ============================================================
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n===== P2 SCREENSHOT 10 - DATA TRAINING TESTING =====")

print("Jumlah data training:", X_train.shape[0])
print("Jumlah data testing:", X_test.shape[0])

print("\nDistribusi target data training:")
print(y_train.value_counts())

print("\nDistribusi target data testing:")
print(y_test.value_counts())


# ============================================================
# 6. PIPELINE PREPROCESSING
# ============================================================
# 6A. PREPROCESSING DATA NUMERIK
# ------------------------------------------------------------
# Mengisi missing value menggunakan median
# Melakukan standardisasi menggunakan StandardScaler

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# ------------------------------------------------------------
# 6B. PREPROCESSING DATA KATEGORIKAL
# ------------------------------------------------------------

# Mengisi missing value menggunakan modus
# Mengubah kategori menjadi numerik menggunakan OneHotEncoder

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

# ------------------------------------------------------------
# 6C. MENGGABUNGKAN PIPELINE PREPROCESSING
# ------------------------------------------------------------
preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_columns),
    ('cat', categorical_transformer, categorical_columns)
])


# ============================================================
# 7. MEMBANGUN MODEL KLASIFIKASI
# ============================================================
# Menentukan dua algoritma klasifikasi
models = {
    'Logistic Regression': LogisticRegression(
        max_iter=2000,
        class_weight='balanced',
        random_state=42
    ),

    'Random Forest': RandomForestClassifier(
        n_estimators=300,
        class_weight='balanced',
        random_state=42,
        n_jobs=-1,
        min_samples_leaf=2
    )
}

# Menyimpan hasil evaluasi setiap model
results = {}

# Menyimpan confusion matrix setiap model
confusion_matrices = {}

# ------------------------------------------------------------
# 7A. TRAINING DAN EVALUASI MODEL
# ------------------------------------------------------------
for model_name, model in models.items():

    print("\n==========================================")
    print("MODEL:", model_name)
    print("==========================================")

    # Menggabungkan preprocessing dengan model
    pipeline = Pipeline(steps=[
        ('preprocessing', preprocessor),
        ('model', model)
    ])

    # Melatih model menggunakan data training
    pipeline.fit(X_train, y_train)

    # Melakukan prediksi menggunakan data testing
    y_pred = pipeline.predict(X_test)

    # Menghitung probabilitas kelas positif
    y_proba = pipeline.predict_proba(X_test)[:, 1]

    # Menghitung metrik evaluasi
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    balanced_acc = balanced_accuracy_score(
        y_test,
        y_pred
    )

    roc_auc = roc_auc_score(
        y_test,
        y_proba
    )

    # --------------------------------------------------------
    # 7B. HASIL EVALUASI MODEL
    # --------------------------------------------------------
    print("\n===== P2 SCREENSHOT 11 - EVALUASI MODEL =====")
    print("Nama Model         :", model_name)
    print("Accuracy           :", round(accuracy * 100, 2), "%")
    print("Precision          :", round(precision * 100, 2), "%")
    print("Recall             :", round(recall * 100, 2), "%")
    print("F1-Score           :", round(f1 * 100, 2), "%")
    print("Balanced Accuracy  :", round(balanced_acc * 100, 2), "%")
    print("ROC-AUC            :", round(roc_auc * 100, 2), "%")

    # Menampilkan classification report
    print("\nClassification Report:")

    print(
        classification_report(
            y_test,
            y_pred,
            target_names=[
                'Existing Customer',
                'Attrited Customer'
            ]
        )
    )

    # --------------------------------------------------------
    # 7C. CONFUSION MATRIX
    # --------------------------------------------------------
    cm = confusion_matrix(y_test, y_pred)

    # Menyimpan confusion matrix
    confusion_matrices[model_name] = cm

    print("\n===== P2 SCREENSHOT 12 - CONFUSION MATRIX =====")
    print("Nama Model:", model_name)
    print(cm)

    display_matrix = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=[
            'Existing Customer',
            'Attrited Customer'
        ]
    )

    display_matrix.plot(cmap='Blues')

    plt.title(f'Confusion Matrix - {model_name}')
    plt.tight_layout()
    plt.show()

    # Menyimpan hasil evaluasi model
    results[model_name] = {
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1-Score': f1,
        'Balanced Accuracy': balanced_acc,
        'ROC-AUC': roc_auc
    }


# ============================================================
# 8. PERBANDINGAN HASIL MODEL
# ============================================================
comparison = pd.DataFrame(results).T

print("\n===== P2 SCREENSHOT 13 - PERBANDINGAN MODEL =====")

# Mengubah hasil evaluasi menjadi persentase
display((comparison * 100).round(2))

# Menentukan model terbaik berdasarkan F1-Score
best_model = comparison['F1-Score'].idxmax()

print("\nModel terbaik berdasarkan F1-Score:", best_model)

# ============================================================
# SELESAI - P2 BANK CUSTOMER CHURN PREDICTION
# ============================================================
```
